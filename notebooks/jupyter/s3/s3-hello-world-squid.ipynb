```python
from squid.resources.s3 import get_s3
```

```python
s3 = get_s3()
s3
```

```python
import boto3
import pandas as pd
bucket_response = s3.list_buckets()

buckets = pd.DataFrame(
    [
        {
            'Name': bucket['Name'],
            'CreationDate': bucket['CreationDate'],
        }
        for bucket in bucket_response.get('Buckets', [])
    ],
    columns=['Name', 'CreationDate'],
)

display(buckets)
```

```python
if buckets.empty:
    print("No buckets available.")
else:
    sample_bucket = buckets.iloc[0]["Name"]
    object_response = s3.list_objects_v2(Bucket=sample_bucket, MaxKeys=10)
    objects = pd.DataFrame(
        [
            {
                "Key": item["Key"],
                "Size": item["Size"],
                "LastModified": item["LastModified"],
            }
            for item in object_response.get("Contents", [])
        ],
        columns=["Key", "Size", "LastModified"],
    )
    print(f"First {len(objects)} object(s) in s3://{sample_bucket}:")
    display(objects)
```
