```python
from squid.resources.postgres import get_cursor
```

```python

def query(sql: str):
    df = (
        spark.read
        .format("postgresql")

        .option("host",      postgres_host)
        .option("port",      postgres_port)
        .option("database",  postgres_database)
        .option("user",      postgres_user)
        .option("password",  postgres_password)
        .option("query",     sql)
        .option("sslmode",   "require")

        .option("connectTimeout", 10)
        .option("numPartitions", 1)

        .load()
    )
    return df
```

```python
df = query("select version()")
display(df)
```

```python
df = query("""
    SELECT
      current_user AS database_user,
      ssl,
      version AS tls_version,
      cipher AS tls_cipher,
      bits AS tls_bits,
      client_dn,
      inet_client_addr()::text AS spark_worker_address,
      pg_backend_pid() AS backend_pid
    FROM pg_stat_ssl
    WHERE pid = pg_backend_pid()
""")
display(df)
```
