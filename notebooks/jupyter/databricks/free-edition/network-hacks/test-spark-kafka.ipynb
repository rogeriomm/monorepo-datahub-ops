```python
from pyspark.sql.utils import AnalysisException

def check_kafka_connector() -> bool:
    """
    Checks whether Spark recognizes 'kafka' as a Structured Streaming source.

    This does not require a working Kafka broker.
    """

    try:
        kafka_df = (
            spark.readStream
            .format("kafka")
            .option("kafka.bootstrap.servers", "localhost:9092")
            .option("subscribe", "__connector_test__")
            .load()
        )

        print("✅ Kafka connector is available.")
        print("Kafka DataFrame schema:")
        kafka_df.printSchema()

        return True

    except Exception as error:
        message = str(error)

        print("❌ Kafka connector could not be loaded.")
        print(f"Error type: {type(error).__name__}")
        print(f"Error: {message}")

        if (
            "DATA_SOURCE_NOT_FOUND" in message
            or "Failed to find the data source: kafka" in message
            or "kafka.DefaultSource" in message
        ):
            print(
                "\nSpark Structured Streaming is available, "
                "but the Spark Kafka connector is missing."
            )

        return False


kafka_supported = check_kafka_connector()
```
