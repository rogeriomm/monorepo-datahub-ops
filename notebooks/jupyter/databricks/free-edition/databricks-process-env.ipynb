# Databricks notebook Python process

The process is started through `sudo`, preserving a large set of environment variables and then switching to a sandboxed Spark user.

## Command structure

```text
sudo
└── preserve selected environment variables
    └── run as user: spark-be603da9-a997-4a11-b246-af
        └── set environment variable:
            DATABRICKS_RUNTIME_VERSION=client.5.9
            └── execute sandbox setup:
                /databricks/python_shell/scripts/sandbox_env_setup.sh
                └── start notebook-scoped Python:
                    /local_disk0/.ephemeral_nfs/envs/pythonEnv-be603da9-a997-4a11-b246-aff71ba92b0b/python_start_notebook_scoped.sh
                    └── launch Python interpreter:
                        /local_disk0/.ephemeral_nfs/envs/pythonEnv-be603da9-a997-4a11-b246-aff71ba92b0b/bin/python
                        └── launch Jupyter kernel:
                            /databricks/python_shell/scripts/db_ipykernel_launcher.py
                            └── use connection file:
                                /local_disk0/sandboxapi/5434141154978540186_1/connection.json
```

## Simplified command

```bash
sudo \
  --preserve-env=<environment-variable-list> \
  -u spark-be603da9-a997-4a11-b246-af \
  env DATABRICKS_RUNTIME_VERSION=client.5.9 \
  /databricks/python_shell/scripts/sandbox_env_setup.sh \
  /local_disk0/.ephemeral_nfs/envs/pythonEnv-be603da9-a997-4a11-b246-aff71ba92b0b/python_start_notebook_scoped.sh \
  /local_disk0/.ephemeral_nfs/envs/pythonEnv-be603da9-a997-4a11-b246-aff71ba92b0b/bin/python \
  /databricks/python_shell/scripts/db_ipykernel_launcher.py \
  -f /local_disk0/sandboxapi/5434141154978540186_1/connection.json
```

## Main components

| Component | Purpose |
|---|---|
| `sudo` | Starts the process under another user |
| `--preserve-env=...` | Keeps selected environment variables |
| `-u spark-be603da9-a997-4a11-b246-af` | Runs the process as the sandbox Spark user |
| `env DATABRICKS_RUNTIME_VERSION=client.5.9` | Sets the runtime version for the child process |
| `sandbox_env_setup.sh` | Configures the Databricks sandbox environment |
| `python_start_notebook_scoped.sh` | Prepares the notebook-scoped Python environment |
| `bin/python` | Notebook virtual-environment Python interpreter |
| `db_ipykernel_launcher.py` | Starts the Databricks Jupyter/IPython kernel |
| `connection.json` | Defines the Jupyter kernel communication ports and credentials |

## Important observation

The value following `SPARK_WARM_UP_CONF_JSON` appears to contain embedded JSON:

```json
{
  "databricks.daemon.driver.lazyConfig.path": "/databricks/secrets/config.json",
  "databricks.loopbackRpc.enabled": true
}
```
