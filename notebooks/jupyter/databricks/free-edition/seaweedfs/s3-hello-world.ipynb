```python
trino_user = dbutils.secrets.get(
    scope="on-premises-integration",
    key="seaweedfs-user",
)

trino_password = dbutils.secrets.get(
    scope="on-premises-integration",
    key="seaweedfs-password",
)
```

```python
import warnings
from urllib3.exceptions import InsecureRequestWarning

warnings.simplefilter("ignore", InsecureRequestWarning)

s3 = boto3.client(
    "s3",
    endpoint_url="https://pub.worldb.dedyn.io:9001",
    aws_access_key_id=trino_user,
    aws_secret_access_key=trino_password,
    region_name="us-east-1",
    #verify=certificate
    verify=False
)

response = s3.list_buckets()
print(response)
print("Buckets:")
for bucket in response["Buckets"]:
    print(bucket["Name"])
```
