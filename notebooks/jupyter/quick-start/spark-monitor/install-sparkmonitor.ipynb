````python
from pyspark.sql import SparkSession
def print_listeners(spark : SparkSession):
    # Access the JVM
    jvm = spark._jvm

    # Get the listener bus
    listener_bus = jvm.org.apache.spark.scheduler.LiveListenerBus

    # Get the SparkContext's listener bus
    sc_listener_bus = spark.sparkContext._jsc.sc().listenerBus()

    # Get the list of registered listeners
    listeners = sc_listener_bus.listeners()

    print("Spark Listeners:")
    print("====================================================")
    # Print each listener
    for listener in listeners:
        print(listener.getClass().getName())
````

````python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
        .getOrCreate()

scala_version = spark.sparkContext._jvm.scala.util.Properties.versionString()
print(f"Scala version: {scala_version}")
print_listeners(spark)
````
