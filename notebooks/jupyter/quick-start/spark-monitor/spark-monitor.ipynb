```python
from pyspark import SparkConf, SparkContext
from pyspark.sql import SparkSession
import os

venv = os.environ['VIRTUAL_ENV']
jar = f"{venv}/lib/python3.13/site-packages/sparkmonitor/listener_2.12.jar"

print(f"Using jar: {jar}")

spark = SparkSession.builder \
    .appName("TestMonitor") \
    .config("spark.extraListeners", "sparkmonitor.listener.JupyterSparkMonitorListener") \
    .config("spark.driver.extraClassPath", jar) \
    .getOrCreate()

# Function to print listeners
def print_listeners(spark):
    listeners = spark.sparkContext._jsc.sc().listenerBus().listeners()
    for i in range(listeners.size()):
        listener = listeners.get(i)
        print(listener.getClass().getName())

print_listeners(spark)


# You can tweak SparkConf as usual:
conf = SparkConf().setAppName("MyApp").setMaster("local[*]")

# Create SparkSession (or SparkContext)
spark = SparkSession.builder.config(conf=conf).getOrCreate()
sc = spark.sparkContext

# Any Spark action will now show the SparkMonitor UI below the cell:
rdd = sc.parallelize(range(1000000)).map(lambda x: x * 2)
print(rdd.sum())
```
