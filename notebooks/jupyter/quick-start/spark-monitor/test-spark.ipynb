```python
from pyspark.sql import SparkSession

def print_listeners(spark : SparkSession):
    # Access the JVM
    jvm = spark._jvm

    # Get the listener bus
    listener_bus = jvm.org.apache.spark.scheduler.LiveListenerBus

    # Get the SparkContext's listener bus
    sc_listener_bus = spark.sparkContext._jsc.sc().listenerBus()

    # Get the list of registered listeners
    listeners = sc_listener_bus.listeners()

    # Print each listener
    for listener in listeners:
        print(listener.getClass().getName())


spark = SparkSession.builder\
        .getOrCreate()

print_listeners(spark)
```

```python
from pyspark import SparkConf, SparkContext
from pyspark.sql import SparkSession

# You can tweak SparkConf as usual:
conf = SparkConf().setAppName("MyApp").setMaster("local[*]")

# Create SparkSession (or SparkContext)
spark = SparkSession.builder.config(conf=conf).getOrCreate()
sc = spark.sparkContext
print_listeners(spark)

# Any Spark action will now show the SparkMonitor UI below the cell:
rdd = sc.parallelize(range(1000000)).map(lambda x: x * 2)
print(rdd.sum())
```
