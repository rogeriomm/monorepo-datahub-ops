```python
%%sql
SELECT OWNER, 
       STREET_NAME,
       COOP_APTS, 
       CURMKTTOT,
       humanize_number(CURMKTTOT) AS CURMKTTOT_ -- Current Market Assessed Total Value

FROM  nyc_property_master
WHERE COOP_APTS < 4     -- Number of residential units 
      AND COOP_APTS > 0
      AND OWNER NOT LIKE "% LLC%"
      AND OWNER NOT LIKE "% INC%"
ORDER BY COOP_APTS DESC, CURMKTTOT DESC
```
