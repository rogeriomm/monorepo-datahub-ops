```python
%%sql
SELECT DISTINCT OWNER, STREET_NAME, GARAGE_AREA
FROM  nyc_property_master
WHERE OWNER NOT LIKE "%LLC%"
ORDER BY OWNER DESC, STREET_NAME DESC
```

```python
%%sql
SELECT DISTINCT owner, STREET_NAME
FROM  nyc_property_master
WHERE OWNER LIKE "%MACHADO %"
ORDER BY owner
```

```python
%%sql
SELECT DISTINCT OWNER, STREET_NAME,COOP_APTS
FROM  nyc_property_master
ORDER BY COOP_APTS DESC
```

```python
%%sql
SELECT OWNER, 
       STREET_NAME,
       COOP_APTS, 
       CURMKTTOT,
       humanize_number(CURMKTTOT) AS CURMKTTOT_ -- Current Market Assessed Total Value

FROM  nyc_property_master
WHERE COOP_APTS < 4     -- Number of residential units 
      AND COOP_APTS > 0
      AND OWNER NOT LIKE "% LLC%"
      AND OWNER NOT LIKE "% INC%"
ORDER BY COOP_APTS DESC, CURMKTTOT DESC
```

```python
%%sql
SELECT OWNER,    
       STREET_NAME,
       COOP_APTS, 
       humanize_number(CURMKTTOT) AS CURMKTTOT_ -- Current Market Assessed Total Value
FROM  nyc_property_master
WHERE OWNER LIKE "% MACHADO %" 
```
