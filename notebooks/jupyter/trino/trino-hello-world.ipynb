```python
import csv

from trino.dbapi import connect
```

```python
connection = connect(
    host="trino",
    port=8080,
    user="dev",
    catalog="memory",
    schema="demo",
    http_scheme="http",
)

cursor = connection.cursor()

cursor.execute("CREATE SCHEMA IF NOT EXISTS memory.demo")
cursor.execute("""
    CREATE TABLE IF NOT EXISTS memory.demo.people (
        id BIGINT,
        name VARCHAR
    )
""")
```

```python
!printf 'id,name\n1,Alice\n2,Bob\n3,Charlie\n' > /tmp/people.csv
!cat /tmp/people.csv
```

```python
with open("/tmp/people.csv", newline="", encoding="utf-8") as file:
    for row in csv.DictReader(file):
        cursor.execute(
            "INSERT INTO memory.demo.people VALUES (?, ?)",
            (int(row["id"]), row["name"]),
        )
```
