```python
from IPython import get_ipython
import pandas as pd

magics = get_ipython().magics_manager.magics

rows = (
    [{"Magic": f"%{name}", "Type": "Line"} for name in magics["line"]]
    + [{"Magic": f"%%{name}", "Type": "Cell"} for name in magics["cell"]]
)

df = (
    pd.DataFrame(rows)
    .sort_values("Magic", key=lambda s: s.str.lstrip("%"))
    .reset_index(drop=True)
)

pd.set_option("display.max_rows", None)
display(df)
```
